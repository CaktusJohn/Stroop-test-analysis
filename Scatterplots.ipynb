Построение графиков связи возраста с другими показателями

```python
mask_covid = data['context'] == 'COVID'  #маски для фильтрации это булев массив из 0 и 1
mask_autism = data['context'] == 'Autism'
mask_rehab = data['context'] == 'Реабилитация'
mask_exoskelet = data['context'] == 'Exoskelet'
mask_Sofia = data['context'] == 'Sofia'
mask_apway = data['context'] == 'apway' 
mask_others = ~(mask_covid | mask_autism | mask_rehab | mask_exoskelet | mask_Sofia)
particular_column = 'old';

masks = (mask_covid, mask_autism, mask_rehab, mask_exoskelet, mask_Sofia, mask_apway, mask_others) 

particular_graphs = (5, 9, 10, 11, 12, 13, 16, 22, 23, 37, 39, 49)
for i in range(4,len(data.columns)-1):
    if (data.columns[i] != particular_column and data.columns[i] != 'person') :
        plt.figure(figsize=(16,9), dpi= 100)
        sns.scatterplot(data=data[mask_Sofia], x='old', y=data.columns[i],color='pink',label='Sofia')
        sns.scatterplot(data=data[mask_apway], x='old', y=data.columns[i],color='black',label='apway')
        sns.scatterplot(data=data[mask_covid], x='old', y=data.columns[i],color='red',label='covid')
        sns.scatterplot(data=data[mask_autism], x='old', y=data.columns[i],color='green',label='autism')
        sns.scatterplot(data=data[mask_rehab], x='old', y=data.columns[i],color='brown',label='rehab')
        sns.scatterplot(data=data[mask_exoskelet], x='old', y=data.columns[i],color="orange",label='exoskelet')
        sns.scatterplot(data=data[~(mask_covid | mask_autism | mask_rehab | mask_exoskelet | mask_Sofia | mask_apway)], x='old', y=data.columns[i], color='blue', label='Others')
        plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))# сдвиг легенды за график
        plt.xlabel('Age, 8 groups for each 10 years (years)')
        
       
        
        plt.show() # чтобы при каждой итерации окна не накапливались. сам график при этом не исчезает
        plt.close()
        
```

```python


particular_graphs = (5, 9, 10, 11, 12, 13, 16, 22, 23, 37, 39, 49)

#для того чтобы в графике old_group группы шли по порядку сделаем отдельный случай
age_groups_order = ['4_7', '8_12', '13_16', '17_21', '22_35', '36_60', '60_PLUS']

#используется для преобразования столбца data['old_group'] в категориальный тип данных с явным порядком категорий.
#ordered=True — указывает, что категории имеют явный порядок, который следует учитывать при построении графиков 
data['old_group'] = pd.Categorical(data['old_group'], categories=age_groups_order, ordered=True)   

color_dict = sns.color_palette("bright", n_colors=7)  # 6 цветов для различных контекстов

for i in range(9, len(data.columns)-1):
    if (data.columns[i] != particular_column and data.columns[i] != 'person' and data.columns[i] != 'Q_common'): 
        
        plt.figure(figsize=(fig_width, fig_height), dpi=300)
        
        
        sns.scatterplot(
        data=data[mask_Sofia], 
        x=data['old'][mask_Sofia] + np.random.uniform(-0.5, 0.5, size=len(data[mask_Sofia])),# передается как данные
        y=data.columns[i],#передаётся как название солбца
        label='София',
        color=color_dict[0]
    )
        
        # последнее значит что точки не рис(уже есть)
        sns.regplot(data=data[mask_Sofia], color=color_dict[0], x='old', y=data.columns[i], line_kws={'linewidth': 2}, scatter=False)

        
        sns.scatterplot(
        data=data[mask_covid], 
        x=data['old'][mask_covid] + np.random.uniform(-0.5, 0.5, size=len(data[mask_covid])), 
        y=data.columns[i],
        label='Ковид',
        color=color_dict[2]
    )
        sns.regplot(data=data[mask_covid],color=color_dict[2], x='old', y=data.columns[i], line_kws={'linewidth': 2}, scatter=False)

        sns.scatterplot(
        data=data[mask_autism], 
        x=data['old'][mask_autism] + np.random.uniform(-0.5, 0.5, size=len(data[mask_autism])), 
        y=data.columns[i],
        label='Аутизм',
        color='black'
    )
        #sns.regplot(ci=None, data=data[mask_autism], color='black', x='old', y=data.columns[i], line_kws={'linewidth': 2}, scatter=False)

        sns.scatterplot(
        data=data[mask_rehab], 
        x=data['old'][mask_rehab] + np.random.uniform(-0.5, 0.5, size=len(data[mask_rehab])), 
        y=data.columns[i], 
        label='Рехаб',
        color='orange'
    )
        sns.regplot(data=data[mask_rehab], color='orange', x='old', y=data.columns[i], line_kws={'linewidth': 2}, scatter=False)

        sns.scatterplot(
        data=data[mask_exoskelet], 
        x=data['old'][mask_exoskelet] + np.random.uniform(-0.5, 0.5, size=len(data[mask_exoskelet])), 
        y=data.columns[i],
        label='Экзоскелет',
        color=color_dict[5]
    )
        sns.regplot(data=data[mask_exoskelet], color=color_dict[5], x='old', y=data.columns[i], line_kws={'linewidth': 2}, scatter=False)

        sns.scatterplot(
        data=data[~(mask_covid | mask_autism | mask_rehab | mask_exoskelet | mask_Sofia)],
        x=data['old'][~(mask_covid | mask_autism | mask_rehab | mask_exoskelet | mask_Sofia) ] + 
        np.random.uniform(-0.5, 0.5, size=len(data[~(mask_covid | mask_autism | mask_rehab | mask_exoskelet | mask_Sofia) ])),
        y=data.columns[i],
        label='Другие',
        color=color_dict[6]
    )
        sns.regplot(
            data=data[~(mask_covid | mask_autism | mask_rehab | mask_exoskelet | mask_Sofia )],
            color=color_dict[6],
            x='old', y=data.columns[i],
            line_kws={'linewidth': 2},
            scatter=False  
            
    )
            
           
        if i == 9:
            plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))# сдвиг легенды за график
            plt.xlabel('Возраст, 8 групп по 10 лет (года)', fontsize=14, fontweight='light')
            plt.ylabel('Продолжительность теста', fontsize=14, fontweight='light')
            plt.title('а', loc='left')
        else:
            plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))# сдвиг легенды за график
            plt.xlabel('Age, 8 groups for each 10 years (years)', fontsize=14, fontweight='light')
            plt.ylabel(data.columns[i], fontsize=14, fontweight='light')
        '''
        #сохраниение
        if (i in particular_graphs):
            column_number = str(i).zfill(2)
            plt.savefig(f'D:/AI/HIstogram/scat_vith_regplot/particular/plot_{column_number}_{data.columns[i]}_vs_{particular_column}.png', format='png', dpi=600)
        else:
            column_number = str(i).zfill(2)
            plt.savefig(f'D:/AI/HIstogram/scat_vith_regplot/plot_{column_number}_{data.columns[i]}_vs_{particular_column}.png', format='png', dpi=600)
        # plt.gca().invert_yaxis()  инвертировать y для олд груп например
        '''
        plt.show()
        plt.close()
```
